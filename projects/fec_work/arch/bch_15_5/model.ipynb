```python
def vector_to_poly(v, gf):
    ak = [ELEMENT_ZERO if e == 0 else ELEMENT_ONE for e in v]
    return GaloisPoly(gf, ak)

def poly_to_vector(p):
    return [0 if ak == ELEMENT_ZERO else 1 for ak in p._p[::-1]]
```

```python
# in list output always interpret:
#     . -1    == ELEMENT_ZERO == 0
#     . 0     == ELEMENT_ONE == 1
#     . 1     == alpha

gf = GaloisField(1, 0b11)
print(gf)

g = GaloisPoly(gf, [ELEMENT_ONE,ELEMENT_ZERO,ELEMENT_ONE,ELEMENT_ZERO,ELEMENT_ZERO,ELEMENT_ONE,ELEMENT_ONE,ELEMENT_ZERO,ELEMENT_ONE,ELEMENT_ONE,ELEMENT_ONE])
print(g)

# Parity == Reminder{ uX**(n-k) / g }, cw == [u | Parity]
n = 15
k = 5

xp  = gf.do_unpack(0b10000000000, bit_width=n-k+1)
xp  = vector_to_poly(xp, gf)

codewords = []
for w in range(1,2**k):
    u = gf.do_unpack(w, bit_width=k)
    u = vector_to_poly(u[::-1], gf)

    print(f"p = {(u * xp) % g} = ({u} * {xp}) % {g} = ({u*xp}) % {g}")
    p = (u * xp) % g
    p = poly_to_vector(p)

    print(f"{poly_to_vector(u)} --- {p}")
    codewords.append(np.hstack((u,p)))

```
